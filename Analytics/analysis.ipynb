### Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
%matplotlib inline
```

### Reading .csv Files

```python
data = pd.read_csv("F:\\Final Year Project\\Datasets\\NCRB-Data.csv", index_col = "State/UT")
data.head()
```

### List of all regions

```python
list1 = data.index.tolist()
list1
```

### Multi Linear Plots

```python
for i in list1:
    df = data.loc[i]
    plot = px.line(df, x="Year", y=['Murders', 'Kidnapping', 'Crimes_against_children', 'Crimes_against_senior_citizen', 'Crimes_against_women', 'Corruption', 'Cyber_Crime', 'Total_Crimes'], labels={'x':'Year','value':'Cases Reported','variable':'Types of Crimes'}, title="Crime Reported in " + i + " During 2014 to 2021", markers=True)
    plot.update_xaxes(ticks="outside", tickwidth=2, tickcolor='crimson', dtick=1)
    plot.update_yaxes(ticks="outside", tickwidth=2, tickcolor='crimson')
    plot.update_layout(font_family="Roboto",font_color="black",title_font_family="Arial Black",title_font_color="black",legend_title_font_color="black")
    plot.write_html(i + '_linear.html')
```

### Stacked Bar Charts

```python
for i in list1:
    df = data.loc[i]
    plot = px.bar(df, x="Year", y=['Murders', 'Kidnapping', 'Crimes_against_children', 'Crimes_against_senior_citizen', 'Crimes_against_women', 'Corruption', 'Cyber_Crime', 'Total_Crimes'], labels={'x':'Year','value':'Cases Reported','variable':'Types of Crimes'}, title="Crime Reported in " + i + " During 2014 to 2021")
    plot.update_xaxes(dtick=1)
    plot.update_yaxes(ticks="outside", tickwidth=2, tickcolor='crimson')
    plot.update_layout(font_family="Roboto",font_color="black",title_font_family="Arial Black",title_font_color="black",legend_title_font_color="black")
    plot.write_html(i + '_stbar.html')
```

### Plotting data on India map using Geopandas

```python
import geopandas as gpd
```

```python
fp = "Igismap/Indian_States.shp"
map_df = gpd.read_file(fp)
map_df.drop([map_df.index[7]])
```

```python
data = pd.read_csv("F:\\Final Year Project\\Datasets\\plot_data.csv")
```

```python
crimes = ['Murders','Kidnapping','Crimes_against_children','Crimes_against_senior_citizen','Crimes_against_women','Corruption','Total_Crimes','Cyber_Crime']
l = [2014,2015,2016,2017,2018,2019,2020,2021]
for j in l:
    for i in crimes:
        data_for_map = data[['State/UT', i, 'Year']]
        data_for_map = data_for_map[data_for_map['Year']==j]
        merged = map_df.set_index('st_nm').join(data_for_map.set_index('State/UT'))
        fig, ax = plt.subplots(1, figsize=(10, 6))
        ax.axis('off')
        ax.set_title(i + ' Reported in India in ' + str(j), fontdict={'fontsize': '25', 'fontweight' : '3'})
        plot = merged.plot(column=i, cmap='YlOrRd', linewidth=0.8, ax=ax, edgecolor='0.8', legend=True)
        fig.savefig(str(j) + i + "_State_wise.png", dpi=100)
```

### Choropleth Plots of India

```python
import pandas as pd
import plotly.graph_objects as go

data = pd.read_csv("F:\\Final Year Project\\Datasets\\plt_data.csv", index_col='Year')
crimes = ['Murders','Kidnapping','Crimes_against_children','Crimes_against_senior_citizen','Crimes_against_women','Corruption','Total_Crimes','Cyber_Crime']
l = [2014,2015,2016,2017,2018,2019,2020,2021]

for i in l:
    for j in crimes:
        df = data.loc[i]
        fig = go.Figure(data=go.Choropleth(geojson="https://gist.githubusercontent.com/jbrobst/56c13bbbf9d97d187fea01ca62ea5112/raw/e388c4cae20aa53cb5090210a42ebb9b765c0a36/india_states.geojson",featureidkey='properties.ST_NM',locationmode='geojson-id',locations=df['State/UT'],z=df[j],autocolorscale=False,colorscale='Reds',marker_line_color='peachpuff',colorbar=dict(title={'text': j},thickness=15,len=0.35,bgcolor='rgba(255,255,255,0.6)',tick0=0,dtick=10000,xanchor='left',x=0.01,yanchor='bottom',y=0.05)))
        fig.update_geos(visible=False,projection=dict(type='conic conformal',parallels=[12.472944444, 35.172805555556],rotation={'lat': 24, 'lon': 80}),lonaxis={'range': [68, 98]},lataxis={'range': [6, 38]})
        fig.update_layout(title=dict(
        text= j + " Reported in India by State in " + str(i),
        xanchor='center',
        x=0.5,
        yref='paper',
        yanchor='bottom',
        y=1,
        pad={'b': 10}
        ),
        margin={'r': 0, 't': 30, 'l': 0, 'b': 0},
        )
        fig.write_html(str(i) + "_"+ j +".html")
```

### Generating Heat maps

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import folium
from folium.plugins import HeatMap

# create a map object
mapObj = folium.Map([24.2170111233401, 81.0791015625000], zoom_start=6)

crimes = ['Murders','Kidnapping','Crimes_against_children','Crimes_against_senior_citizen','Crimes_against_women','Corruption','Total_Crimes','Cyber_Crime']
l = [2014,2015,2016,2017,2018,2019,2020,2021]

data = pd.read_csv("F:\\Final Year Project\\Datasets\\plt_data.csv")
lat_long = pd.read_csv("F:\\Final Year Project\\Datasets\\lat_long.csv")

scaler = MinMaxScaler()
df_scaled = scaler.fit_transform(data[['Murders','Kidnapping','Crimes_against_children','Crimes_against_senior_citizen','Crimes_against_women','Corruption','Total_Crimes','Cyber_Crime']].to_numpy())
df_scaled = pd.DataFrame(df_scaled, columns=['Murders','Kidnapping','Crimes_against_children','Crimes_against_senior_citizen','Crimes_against_women','Corruption','Total_Crimes','Cyber_Crime'])
data.drop(['Murders','Kidnapping','Crimes_against_children','Crimes_against_senior_citizen','Crimes_against_women','Corruption','Total_Crimes','Cyber_Crime'],axis=1,inplace=True)
data = pd.concat([data,df_scaled],axis=1)
```

```python
for i in l:
    for j in crimes:
        data_for_map = data[['State/UT', j, 'Year']]
        data_for_map = data_for_map[data_for_map['Year']==i]
        merged = lat_long.set_index('State/UT').join(data_for_map.set_index('State/UT'))
        merged.drop(['Year'],axis=1,inplace=True)
        merged.reset_index()
        merged = merged[['lat','long',j]]
        HeatMap(merged).add_to(mapObj)
        mapObj.save(str(i) + "_" + j + ".html")
        
```
